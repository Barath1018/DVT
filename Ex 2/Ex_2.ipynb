```python
import sqlite3
import pandas as pd

conn = sqlite3.connect(":memory:")
cursor = conn.cursor()

df = pd.read_csv("Social_Media_Engagement.csv")

df.to_sql(
    "Social_Media_Engagement",
    conn,
    if_exists="replace",
    index=False
)

tables = pd.read_sql_query(
    "SELECT name FROM sqlite_master WHERE type='table';",
    conn
)

print("Tables in Database:")
print(tables)

# Get first table name
table_name = tables.iloc[0, 0]

# Read data
df = pd.read_sql_query(f"SELECT * FROM {table_name}", conn)

print("\n===== HEAD =====")
print(df.head())

print("\n===== INFO =====")
df.info()

conn.close()
```

```python
df.to_csv("Social_Media_Engagement_export.csv", index=False)

print("CSV exported successfully!")
```

```python
df.to_excel("Social_Media_Engagement_export.xlsx", index=False)

print("Excel file exported successfully!")
```

```python
import sqlite3

conn = sqlite3.connect("Social_Media_Engagement_export.db")

df.to_sql(
    "mytable",              # Table name
    conn,
    if_exists="replace",    # replace existing table
    index=False
)

print("Data exported successfully to SQLite database!")

# Close connection
conn.close()
```
