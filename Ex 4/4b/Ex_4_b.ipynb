```python
import pandas as pd
df = pd.read_csv("Social_Media_Engagement.csv")
print(df)
print(df.head())
print(df.info())
print(df.describe())
```

```python
import pandas as pd


print("===== SHAPE =====")
print(df.shape)


print("\n===== DATA TYPES =====")
print(df.dtypes)


print("\n===== FILTER COLUMNS =====")
print(df[["Username", "Platform", "Follower Count"]])


print("\n===== FOLLOWER COUNT > 50000 =====")
print(df[df["Follower Count"] > 50000])


print("\n===== PLATFORM = Instagram =====")
print(df[df["Platform"] == "Instagram"])


print("\n===== PLATFORM = YouTube =====")
print(df[df["Platform"] == "YouTube"])


# Multiple Conditions
print("\n===== FOLLOWER COUNT > 50000 AND PLATFORM = Instagram =====")
print(df[(df["Follower Count"] > 50000) & (df["Platform"] == "Instagram")])


print("\n===== SORT BY FOLLOWER COUNT =====")
print(df.sort_values(by="Follower Count"))


print("\n===== SORT BY FOLLOWER COUNT DESC =====")
print(df.sort_values(by="Follower Count", ascending=False))


print("\n===== SORT BY PLATFORM AND FOLLOWER COUNT =====")
print(df.sort_values(by=["Platform", "Follower Count"]))
```

```python
print(df["Follower Count"].corr(df["Posts Per Week"]))

print(df["Engagement Rate"].corr(df["Conversion Rate"]))
```

```python
print(df.corr(numeric_only=True, method='pearson'))

print(df.corr(numeric_only=True, method='spearman'))
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# =====================================
# View Data
# =====================================

print("First 5 Rows")
print(df.head())

print("\nShape")
print(df.shape)

print("\nData Types")
print(df.dtypes)

# =====================================
# Filter Data
# =====================================

print("\nAccounts with Follower Count > 50000")
follower_filter = df[df["Follower Count"] > 50000]
print(follower_filter)

print("\nAccounts on Instagram")
instagram_accounts = df[df["Platform"] == "Instagram"]
print(instagram_accounts)

print("\nAccounts on YouTube")
youtube_accounts = df[df["Platform"] == "YouTube"]
print(youtube_accounts)

# =====================================
# Sort Data
# =====================================

print("\nSorted by Follower Count")
sorted_df = df.sort_values(by="Follower Count", ascending=False)
print(sorted_df)

# =====================================
# Statistics
# =====================================

print("\nMean Follower Count =", df["Follower Count"].mean())
print("Median Follower Count =", df["Follower Count"].median())

print("\nMode Follower Count")
print(df["Follower Count"].mode())

print("Standard Deviation Follower Count =", df["Follower Count"].std())

# =====================================
# Summary Statistics
# =====================================

print("\nSummary Statistics")
print(df.describe())

# =====================================
# Group By
# =====================================

print("\nAverage Follower Count by Platform")
platform_avg = df.groupby("Platform")["Follower Count"].mean()
print(platform_avg)

# =====================================
# Correlation
# =====================================

print("\nCorrelation")
corr = df[["Follower Count", "Engagement Rate"]].corr()
print(corr)

# =====================================
# Scatter Plot
# =====================================

plt.figure(figsize=(6,4))
plt.scatter(df["Follower Count"], df["Engagement Rate"])
plt.title("Follower Count vs Engagement Rate")
plt.xlabel("Follower Count")
plt.ylabel("Engagement Rate")
plt.grid(True)
plt.show()

# =====================================
# Bar Plot
# =====================================

platform_avg.plot(kind="bar")
plt.title("Average Follower Count by Platform")
plt.xlabel("Platform")
plt.ylabel("Average Follower Count")
plt.show()
```
